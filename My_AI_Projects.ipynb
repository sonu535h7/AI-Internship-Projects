```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# 1. We create a tiny list of fake emails to teach the computer
emails = [
    "Get free money now! Click this link for 1000 dollars!",
    "Hey friend, are we still playing video games tonight?",
    "WINNER! You won a free iPad. Call this number now!",
    "Hi Mom, I will be home late for dinner today."
]

# 1 means Spam (bad). 0 means Ham (good).
labels = [1, 0, 1, 0]

# 2. Teach the computer what words look like
vectorizer = CountVectorizer()
email_numbers = vectorizer.fit_transform(emails)

# 3. Train the AI model brain
ai_brain = MultinomialNB()
ai_brain.fit(email_numbers, labels)

print("Level 1 Complete: The AI is now trained!")
```

```python
# Change the words inside the quotes below to whatever you want!
my_test_email = "Congratulations! You won a free holiday vacation trip!"

# Let the AI guess
new_numbers = vectorizer.transform([my_test_email])
prediction = ai_brain.predict(new_numbers)

if prediction[0] == 1:
    print("🚨 WARNING: This looks like SPAM!")
else:
    print("✅ SAFE: This looks like a normal email.")
```

```python
import urllib.request

# This downloads a quick 2-second audio file of someone speaking from the internet
audio_url = "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-1.mp3"
urllib.request.urlretrieve(audio_url, "sample.mp3")

print("Audio file safely downloaded!")
```

```python
import speech_recognition as sr

# 1. Set up the speech reader
recognizer = sr.Recognizer()

# 2. Open the audio file we just downloaded
with sr.AudioFile("sample.wav") as source:
    audio_data = recognizer.record(source)

# 3. Ask Google to translate the sound to text
try:
    text = recognizer.recognize_google(audio_data)
    print("🤖 The Robot Hears: " + text)
except:
    print("Oops, could not understand the audio!")
```
